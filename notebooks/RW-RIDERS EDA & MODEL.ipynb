### Customer Churn Definition
#### RECENCY

```python
import pandas as pd

df['pickup_time'] = pd.to_datetime(df['pickup_time'])

last_date = df['pickup_time'].max()

customer_last_trip = df.groupby('user_id')['pickup_time'].max().reset_index()

customer_last_trip['days_since_last_trip'] = (
    last_date - customer_last_trip['pickup_time']
).dt.days

customer_last_trip['churn'] = customer_last_trip['days_since_last_trip'].apply(
    lambda x: 1 if x > 30 else 0
)
```

```python
customer_last_trip
```

### Driver Churn Definition

```python
driver_last_trip = df.groupby('driver_id')['pickup_time'].max().reset_index()

driver_last_trip['days_since_last_trip'] = (
    last_date - driver_last_trip['pickup_time']
).dt.days

driver_last_trip['churn'] = driver_last_trip['days_since_last_trip'].apply(
    lambda x: 1 if x > 30 else 0
)
```

```python
driver_last_trip
```

#### FREQUENCY

```python
customer_freq = df.groupby('user_id')['trip_id'].count().reset_index()
customer_freq.rename(columns={'trip_id': 'total_trips'}, inplace=True)
```

```python
customer_freq.head(5)
```

#### MONETARY

```python
customer_monetary = df.groupby('user_id')['fare'].sum().reset_index()
customer_monetary.rename(columns={'fare': 'total_spend'}, inplace=True)
```

```python
customer_monetary.head(5)
```

#### LOCATION BEHAVIOR

```python
customer_city = df.groupby('user_id')['city'].agg(lambda x: x.mode()[0]).reset_index()
customer_city.head(5)
```

```python
customer_avg_surge = df.groupby('user_id')['surge_multiplier'].mean().reset_index()
```

```python
X_numeric = X_numeric.fillna(0)   # simple approach
# OR
# X_numeric = X_numeric.dropna()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_numeric, y, test_size=0.2, random_state=42, stratify=y
)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

log_model = LogisticRegression(max_iter=1000)
log_model.fit(X_train, y_train)

rf_model = RandomForestClassifier(n_estimators=200, max_depth=10, random_state=42)
rf_model.fit(X_train, y_train)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    random_state=42
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)
rf_proba = rf_model.predict_proba(X_test)[:,1]
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

fpr_rf, tpr_rf, _ = roc_curve(y_test, rf_proba)
auc_rf = roc_auc_score(y_test, rf_proba)

plt.figure(figsize=(8,6))
plt.plot(fpr_rf, tpr_rf, label=f'Random Forest AUC: {auc_rf:.2f}')
plt.plot([0,1],[0,1],'--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()
```
