### Convert the pickup time and droptime to time series 

```python
df["pickup_time"] = pd.to_datetime(df["pickup_time"],errors="coerce",utc=True)

df["dropoff_time"] = pd.to_datetime(df["dropoff_time"],errors="coerce",utc=True)
```

```python
df[['pickup_time','dropoff_time']].dtypes
```

```python
df["P_Year"] = df["pickup_time"].dt.year
df["P_Month_Name"] = df["pickup_time"].dt.month_name()
df["P_Quarter"] = df["pickup_time"].dt.quarter
df["P_day"] = df["pickup_time"].dt.day_name()
```

```python
df["D_Year"] = df["dropoff_time"].dt.year
df["D_Month_Name"] = df["dropoff_time"].dt.month_name()
df["D_Quarter"] = df["dropoff_time"].dt.quarter
df["D_day"] = df["dropoff_time"].dt.day_name()
```

### Handing Outliers

```python
numerical_columns = df.select_dtypes(include= "number")
```

```python
len(numerical_columns.columns)
```

```python
fig, ax= plt.subplots(nrows=4, ncols=3, figsize=(20,20))
ax = ax.flatten()

for idx, col in enumerate(numerical_columns):
    sns.histplot(df[col], ax=ax[idx])
    ax[idx].set_title(f"Histoplot for {col}")

plt.show()
```

### Univariate Analysis

```python
fig, ax = plt.subplots(nrows=1, ncols = 2,figsize=(9,4))
ax = ax.flatten()
sns.countplot(x='P_Year', data=df, ax=ax[0])
sns.countplot(x='D_Year', data=df, ax=ax[1])
ax[0].set_title("Pickups by Year")
ax[1].set_title("Dropoffs by Year")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(25,10))
sns.countplot(x='P_Month_Name', data=df)
plt.title("Trips by Month")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(25,10))
ax=sns.countplot(x='P_day', data=df, hue= "P_day")
for container in ax.containers:
    ax.bar_label(container, fmt="%.2f", padding=3)
plt.title("Trips by Day")

plt.tight_layout()
plt.show()
```
